# 01. Data Understanding

This notebook inspects the dataset before modeling. We want to confirm the shape, target variable, missing values, and any structural issues that can affect the downstream analysis.

```python
from pathlib import Path
import pandas as pd

data_path = Path('../data/data.csv')
df = pd.read_csv(data_path)
df.head()
```

```python
print(f"rows={df.shape[0]}")
print(f"columns={df.shape[1]}")
print(df.columns.tolist())
print(df.dtypes.head())
```

```python
print(df['diagnosis'].value_counts())
print(f"missing_values={df.isna().sum().sum()}")
print(f"duplicate_rows={df.duplicated().sum()}")
print(f"unique_ids={df['id'].nunique()}")
print(df.columns[df.isna().any()].tolist())
```

## Interpretation

The dataset has 569 records and 33 columns. The target variable is `diagnosis` with two classes: B and M. We also found one empty trailing column named `Unnamed: 32`, which indicates a CSV export artifact. The `id` column is unique for each row and should not be used as a predictive feature.
